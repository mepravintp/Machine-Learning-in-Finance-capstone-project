## 1. Rank by RAROC

RAROC = (expected return − expected loss) ÷ capital charge, computed in each handoff record. Capital charges: corporate 100%, retail 75%, equity 100%, liquidity reserve 0%.

## 2. Linear programme

`scipy.optimize.linprog`: maximise total risk-adjusted return subject to reserve ≥ the Treasury requirement; each line ≤ its capacity; no line above 35% and HFT at most 3%; combined stress loss ≤ an agreed share of ₹1,000 cr.

## 3. Stress test

One combined scenario: PDs × 1.5, equity −25%, fraud losses × 2, a deposit outflow equal to the worst gap in the file, initiative success rates down 15 points. Report the loss and the remaining liquidity.

## 4. Sensitivity

How the allocation moves when LGD, the terminal value or trading costs change, and which slices are stable.

## 5. Board table

One allocation table, one stress result, the three risks we accept and their controls, and what we chose not to deploy and why. The allocation must add to ₹1,000 cr.

## Traps to check before handing over
